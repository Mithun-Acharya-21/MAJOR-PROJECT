# 02 — Facial Expression Data Exploration

Explore facial expression datasets (FER-2013, RAF-DB, CK+) to understand:
- Image dimensions and quality
- Emotion class distribution
- Sample visualizations
- Preprocessing pipeline verification

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from pathlib import Path
import sys
sys.path.insert(0, '..')

# Count images per class
data_dir = Path('../data/processed/face/train')
if data_dir.exists():
    classes = [d.name for d in data_dir.iterdir() if d.is_dir()]
    for cls in sorted(classes):
        n = len(list((data_dir / cls).glob('*.*')))
        print(f'{cls}: {n} images')
else:
    print(f'Data directory not found: {data_dir}')
    print('Please process raw data first.')
```

```python
# Display sample images per class
fig, axes = plt.subplots(2, 5, figsize=(15, 6))
for i, cls in enumerate(classes[:2]):
    class_dir = data_dir / cls
    images = list(class_dir.glob('*.jpg'))[:5]
    for j, img_path in enumerate(images):
        img = Image.open(img_path)
        axes[i, j].imshow(img)
        axes[i, j].set_title(cls)
        axes[i, j].axis('off')
plt.tight_layout()
plt.show()
```

```python
# Image dimension analysis
from ml.face_pipeline.preprocessing import get_inference_transform

transform = get_inference_transform()
sample_images = list((data_dir / classes[0]).glob('*.jpg'))[:10]

for img_path in sample_images[:3]:
    img = Image.open(img_path)
    tensor = transform(img)
    print(f'{img_path.name}: {img.size} → tensor {tensor.shape}')
```
